```python
from scipy.stats import gaussian_kde
def scatter_with_gaussian_kde(ax, x, y, size):
    xy = np.vstack([x, y])
    z = gaussian_kde(xy)(xy)
    ax.scatter(x, y, c=z, s=size)
```

# Model comparison

```python
topt_test = pd.read_csv('../data/Topt/test.csv');
pred_seq2topt = pd.read_csv('../data/benchmark/pred_seq2topt.csv');
pred_seq2topt['topt'] = list(topt_test.topt)
ptem_out = pd.read_csv('../data/benchmark/test_preoptem.fasta.pred',sep='\t');
ptem_pred = pd.DataFrame(zip(topt_test.uniprot_id, topt_test.topt,ptem_out.Predicted_optimal_temperature ),
                        columns=['ID','topt','pred_topt'])
pred_ET= pd.read_csv('../data/benchmark/DeepET_result.csv')
```

```python
topt_data = pd.read_csv('../data/Topt/sequence_ogt_topt.csv')
def get_ogt(protid, data):
    temp = data[data['uniprot_id']==protid]
    return list(temp['ogt'])[0]
tomer_preds = []
for i in range(len(topt_test.index)):
    prot_id = list(topt_test['uniprot_id'])[i]
    seq = list(topt_test['sequence'])[i]
    y_pred, y_err = tomer.pred_seq_topt(seq, get_ogt(prot_id, topt_data ))
    tomer_preds.append(y_pred)
pred_tomer = pd.DataFrame(zip(list(topt_test['uniprot_id']), list(topt_test['topt']), tomer_preds),
                        columns =['uniprot_id','topt','pred_topt'] )
```

```python
print(get_rmse( pred_seq2topt['topt'], pred_seq2topt['pred_topt']))
print(get_mae( pred_seq2topt['topt'], pred_seq2topt['pred_topt']))
print(get_r2( pred_seq2topt['topt'], pred_seq2topt['pred_topt']))
```

```python
print(get_r2( pred_tomer['topt'], pred_tomer['pred_topt']))
print(get_r2( pred_seq2topt['topt'], pred_seq2topt['pred_topt']))
print(get_r2( pred_seq2topt['topt'], pred_seq2topt['pred_topt']))
print(get_r2( pred_seq2topt['topt'], pred_seq2topt['pred_topt']))
```

```python
pred60 = pd.DataFrame(zip(list(topt_test.topt),list(pred_tomer['pred_topt']), list(ptem_pred['pred_topt']),
                         list(pred_ET['pred_topt']), list(pred_seq2topt['pred_topt']) ),
                     columns=['Exp','TOMER','Preoptem','DeepET','Seq2Topt'])
pred60 = pred60[pred60.Exp>60].reset_index()
```

```python
fig, axs = plt.subplots(1,3, figsize=(18,5))
models = ['TOMER','Preoptem','DeepET','Seq2Topt']
scatter_with_gaussian_kde(axs[0], pred_seq2topt['topt'], pred_seq2topt['pred_topt'], size = 50);
axs[0].plot([10,100],[10,100],'r--');
axs[0].text(x=10,y=100,s='RMSE=12.26\u2103',fontsize=12)
axs[0].text(x=10,y=90,s='MAE=8.89\u2103',fontsize=12)
axs[0].text(x=10,y=80,s='R2=0.57',fontsize=12)
axs[0].set_xlabel('Experimental Topt (\u2103)',fontsize=12)
axs[0].set_ylabel('Predicted Topt (\u2103)',fontsize=12)
axs[0].set_title('Seq2Topt prediction',fontsize=12)

R2_values = [get_r2( pred_tomer['topt'], pred_tomer['pred_topt']),get_r2( ptem_pred['topt'], ptem_pred['pred_topt']),
            get_r2( pred_ET['topt'], pred_ET['pred_topt']), get_r2( pred_seq2topt['topt'], pred_seq2topt['pred_topt'])]
bars = axs[1].bar(models, R2_values,
          color=['tab:brown','orange','tab:purple' ,'tab:red'], width=0.5)
axs[1].set_xlabel('Model',fontsize=12); axs[1].set_ylabel('R2',fontsize=12);
axs[1].set_title('Model comparison on hold-out test set',fontsize=12)
for bar in bars:
    axs[1].text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
    f'{bar.get_height():.2f}', ha='center', va='bottom', fontsize=12)
axs[1].set_ylim([0,0.6])

rmse_60 = [];
for model in models:
    rmse_60.append( get_rmse(pred60.Exp, pred60[model]) )
bars = axs[2].bar(models, rmse_60 ,
          color=['tab:brown','orange','tab:purple' ,'tab:red'], width=0.5)
for bar in bars:
    axs[2].text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
    f'{bar.get_height():.2f}', ha='center', va='bottom', fontsize=12)
axs[2].set_ylim([0,24.0])
axs[2].set_title('Model comparison (Topt>60\u2103)',fontsize=12)
axs[2].set_xlabel('Model',fontsize=12); axs[2].set_ylabel('RMSE (\u2103)',fontsize=12);
# plt.savefig('../../../Topt_paper/Figures/model_eval.svg',format='svg',dpi=150)
```

# Different enzyme classes

```python
pred_result = pd.DataFrame(zip(list(topt_test.uniprot_id),list(topt_test.topt),list(pred_tomer['pred_topt']), 
                list(ptem_pred['pred_topt']),list(pred_ET['pred_topt']), list(pred_seq2topt['pred_topt']) ),
                     columns=['uniprot_id','Exp','TOMER','Preoptem','DeepET','Seq2Topt'])
enzinfo= pd.read_csv('../data/Topt/topt_enzinfo.csv')
```

```python
def get_ec_number(uniprot_id):
    url = f"https://rest.uniprot.org/uniprotkb/{uniprot_id}.json"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        ec_numbers = []
        for activity in data.get("proteinDescription", {}).get("recommendedName", {}).get("ecNumbers", []):
            ec_numbers.append(activity)
        return ec_numbers
    else:
        print(f"Failed to fetch data for UniProt ID {uniprot_id}. Status code: {response.status_code}")
        return None
```

```python
ec_class = []
for pid in list(topt_test.uniprot_id):
    ec_number = list(enzinfo[enzinfo['uniprot_id']==pid]['ec'])[0]
    if ec_number=='unknown':
        ec_class.append('unknown')
    else:
        ec_class.append('EC'+ec_number[0])
pred_result['EC'] = ec_class
pred_result = pred_result[pred_result.EC != 'unknown'].reset_index()
```

```python
fig, ax = plt.subplots(1,1,figsize=(14,5))
X_axis = np.arange(1,7)
xlabels = ['EC'+str(x) for x in X_axis]
rmse_s2, rmse_et, rmse_tom = [],[],[]
for ec in xlabels:
    temp_pd = pred_result[pred_result.EC==ec]
    rmse_s2.append( get_rmse(temp_pd['Exp'],temp_pd['Seq2Topt']) )
    rmse_et.append( get_rmse(temp_pd['Exp'],temp_pd['DeepET']) )
    rmse_tom.append( get_rmse(temp_pd['Exp'],temp_pd['TOMER']) )
ax.bar(X_axis - 0.2, rmse_tom, 0.2, color='tab:brown',label = 'TOMER') 
ax.bar(X_axis - 0.0, rmse_et, 0.2, color='tab:purple',label = 'DeepET') 
ax.bar(X_axis + 0.2, rmse_s2, 0.2, color='tab:red',label = 'Seq2Topt')
ax.set_ylabel('RMSE (\u2103)',fontsize=12)
ax.set_xticks(X_axis, xlabels);
ax.legend()
ax.set_title('Model comparison on different enzyme classes',fontsize=12)
# plt.savefig('../../../Topt_paper/Figures/comparison_ECs.svg',format='svg',dpi=150)
```

## Tomer but OGT values shuffled

```python
ogts = []
for i in range(len(topt_test.index)):
    prot_id = list(topt_test['uniprot_id'])[i]
    ogts.append( get_ogt(prot_id, topt_data ) )
topt_test['ogt'] = ogts
shuffled_ogts = sample(ogts, len(ogts))
```

```python
ogtsf_preds = []
for i in range(len(topt_test.index)):
    seq = list(topt_test['sequence'])[i]
    y_pred, y_err = tomer.pred_seq_topt(seq, shuffled_ogts[i] )
    ogtsf_preds.append(y_pred)
```

```python
print( get_rmse( topt_test['topt'], ogtsf_preds) )
print( get_mae( topt_test['topt'], ogtsf_preds) )
print( get_r2( topt_test['topt'], ogtsf_preds) )
```

```python
fig, ax = plt.subplots(figsize=(6,5.5))
scatter_with_gaussian_kde(ax, topt_test['topt'], ogtsf_preds, size = 30)
ax.plot([10,100],[10,100],'r--')
ax.set_xlim([10,100]);ax.set_ylim([10,100])
ax.set_title('TOMER (OGTs shuffled)')
ax.set_xlabel('Experimental Topt (\u2103)',fontsize=14)
ax.set_ylabel('Predicted Topt (\u2103)',fontsize=14)
ax.text(x=40,y=25,s='RMSE=26.11\u2103')
ax.text(x=40,y=20,s='MAE=19.49\u2103')
ax.text(x=40,y=15,s='R2=-0.96')
```
